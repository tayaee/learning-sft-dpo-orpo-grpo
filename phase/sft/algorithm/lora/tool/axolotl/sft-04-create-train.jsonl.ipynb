```python
from datasets import load_dataset
import pandas as pd

df = load_dataset("amphora/korean-reasoning-small", split="train").to_pandas()
df
```

```python
from tqdm.auto import tqdm
msg = {"messages": []}
for _,row in tqdm(df.iterrows(), total=len(df)):
    msg["messages"].append([
        {
            "role": "user",
            "content": row.instruction
        },
        {
            "role": "assistant",
            "content": row.response
        }
    ])
ds = pd.DataFrame(msg)
```

```python
ds.to_json("train.jsonl", lines=True, orient="records", force_ascii=False)
```
